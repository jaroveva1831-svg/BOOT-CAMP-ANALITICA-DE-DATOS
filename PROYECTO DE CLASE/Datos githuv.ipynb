# 🎲 Mini-reto 11 | Limpieza, Filtrado y Reorganización de Datos Energéticos con Enriquecimiento de Nombres Oficiales

## 🧩 Contexto del Problema

En los estudios sobre calidad del servicio energético en zonas no interconectadas, es frecuente trabajar con bases de datos que contienen inconsistencias en el formato de los valores, caracteres especiales en los nombres de las variables y duplicados en la información reportada por departamentos y municipios. Antes de realizar cualquier análisis comparativo, es indispensable limpiar estas inconsistencias y reorganizar la estructura de los datos.

**Adicionalmente**, los datos originales suelen contener **códigos DANE** en lugar de los nombres completos de departamentos, municipios y centros poblados, o presentan nombres no estandarizados. Por lo tanto, es necesario **enriquecer la tabla principal** con los nombres oficiales provenientes de la tabla **DIVIPOLA** del DANE, asegurando que los análisis territoriales utilicen la nomenclatura oficial.

En este mini-reto trabajarás con un archivo CSV que contiene valores de energía activa, energía reactiva y potencia máxima para diferentes departamentos, municipios y localidades (centros poblados) del país, reportados en varios años. Adicionalmente, contarás con la tabla **DIVIPOLA** que contiene los códigos y nombres oficiales de departamentos, municipios y centros poblados.

Se requiere:
1. **Enriquecer** la tabla principal con los nombres oficiales desde la tabla DIVIPOLA.
2. **Limpiar** las columnas numéricas.
3. **Estandarizar** los nombres de los departamentos y municipios (eliminar tildes).
4. **Filtrar** únicamente el territorio continental (excluir San Andrés).
5. **Reorganizar** los datos en una tabla pivote que resuma la energía activa total por departamento y por año de servicio.

Este ejercicio permite integrar cuatro habilidades prácticas: **relación de tablas (lookup), limpieza de datos, filtrado y transformación estructural mediante pivot tables**, todas esenciales en entornos de análisis energético territorial.

## 🎯 Objetivos

- **Relacionar tablas**: Enriquecer la tabla de consumo energético con los nombres oficiales de departamentos, municipios y centros poblados desde la tabla DIVIPOLA.
- **Limpiar columnas numéricas** que vienen con comas y convertirlas correctamente a tipo numérico.
- **Estandarizar caracteres especiales** en los nombres de departamentos y municipios (eliminar tildes).
- **Filtrar** únicamente los registros del territorio continental (excluir San Andrés y todas sus variantes).
- **Agrupar y sumar** energía activa y energía reactiva por departamento y municipio.
- **Construir una tabla pivote** que muestre la energía activa total por departamento y año.
- **Visualizar** los resultados mediante un gráfico de barras horizontal.

## 📋 Criterios de Calidad

✅ **Enriquecimiento correcto** de la tabla principal con nombres oficiales desde DIVIPOLA usando merges (left joins) por código.

✅ **Conversión correcta** de columnas numéricas (`ENERGÍA ACTIVA`, `ENERGÍA REACTIVA`, `POTENCIA MÁXIMA`) removiendo comas y aplicando `astype()`.

✅ **Normalización de caracteres especiales** en nombres de departamentos y municipios (reemplazo de tildes).

✅ **Aplicación del filtro** que excluye el Archipiélago de San Andrés en todas sus variantes.

✅ **Agrupación adecuada** con `groupby()` y suma de valores numéricos.

✅ **Uso de `pivot_table()`** para reorganizar la energía activa por departamento y año.

✅ **Visualización con matplotlib** usando barra horizontal (`barh`), etiquetas claras y `tight_layout()`.

```python
import pandas as pd

ruta_zni = 'https://github.com/jaroveva1831-svg/BOOT-CAMP-ANALITICA-DE-DATOS/raw/refs/heads/main/Estado_de_la_prestaci%C3%B3n_del_servicio_de_energ%C3%ADa_en_Zonas_No_Interconectadas_20260422.csv'
ruta_divipola = 'https://github.com/jaroveva1831-svg/BOOT-CAMP-ANALITICA-DE-DATOS/raw/refs/heads/main/DIVIPOLA_-_C%C3%B3digos_cabeceras_-_Centros_poblados_20260429.csv'

df_zni = pd.read_csv(ruta_zni)
df_divipola = pd.read_csv(ruta_divipola)

```

```python
display(df_zni.head(3))
display(df_divipola.head(3))
```

```python
#TRAER EL NOMBRE OFICIAL DEL DEPARTAMENTO
df_zni = df_zni.merge(
    df_divipola[['Código Departamento','Nombre_departamento']].drop_duplicates('Código Departamento'),left_on='ID DEPATAMENTO', right_on = 'Código Departamento',how='left'
)

df_zni['DEPARTAMENTO'] = df_zni['Nombre_departamento']
df_zni = df_zni.drop(columns=['Código Departamento','Nombre_departamento'])
```

```python
df_zni.head(50)
```

```python
df_zni['DEPARTAMENTO'].value_counts()
```

```python
df_zni.columns
```

```python
#TRAER EL NOMBRE OFICIAL DEL MUNICIPIO
df_zni = df_zni.merge(
    df_divipola[['Codigo Municipio','Nombre Municipio']].drop_duplicates('Codigo Municipio'),left_on='ID MUNICIPIO', right_on = 'Codigo Municipio',how='left'
)

df_zni['MUNICIPIO'] = df_zni['Nombre Municipio']
df_zni = df_zni.drop(columns=['Codigo Municipio','Nombre Municipio'])
```

```python
df_zni.head(2)
```

```python
#TRAER EL NOMBRE OFICIAL DE LOCALIDAD
df_zni = df_zni.merge(
    df_divipola[['Código Centro Poblado','Nombre Centro Poblado']].drop_duplicates('Código Centro Poblado'),left_on='ID LOCALIDAD', right_on = 'Código Centro Poblado',how='left'
)

df_zni['LOCALIDAD'] = df_zni['Nombre Centro Poblado']
df_zni = df_zni.drop(columns=['Código Centro Poblado','Nombre Centro Poblado'])

##========================
#CORREGIR NOMBRES DE LOS DIAS
df_zni['DÍA DE DEMANDA MÁXIMA'] = df_zni['DÍA DE DEMANDA MÁXIMA'].str.upper()

viejos = ['MIERCOLES','SABADO']
nuevos = ['MIÉRCOLES','SÁBADO']

## CREAR UN DICCIONARIO CON LAS LISTAS EMPAQUETADAS
reemplazos = dict(zip(viejos,nuevos))

###REEMPLAZAR USANDO LOS DICCIONARIOS
df_zni['DÍA DE DEMANDA MÁXIMA'] = df_zni['DÍA DE DEMANDA MÁXIMA'].replace(reemplazos,regex=True)

#CORREGIR PROBLEMAS CON COLUMNAS NUMERICAS QUE SON OBJECT

df_zni['ENERGÍA ACTIVA'] = df_zni['ENERGÍA ACTIVA'].str.replace('.','', regex=False).astype(int)
df_zni['ENERGÍA REACTIVA'] = df_zni['ENERGÍA REACTIVA'].str.replace('.','', regex=False).astype(int)

df_zni.dropna(inplace = True)
df_zni['POTENCIA MÁXIMA'] = df_zni['POTENCIA MÁXIMA'].str.replace('.','', regex=False)
df_zni['POTENCIA MÁXIMA'] = df_zni['POTENCIA MÁXIMA'].str.replace(',','.', regex=False).astype(float).astype(int)



#AGRUPAR DATOS POR POR DEPARTAMENTO Y POR MUNICIPIO
#SUMANDO LOS VALORES DE ENERGIA ACTIVA Y REACTIVA

df_agrupado = df_zni.groupby(['DEPARTAMENTO','MUNICIPIO'])[['ENERGÍA ACTIVA','ENERGÍA REACTIVA']].sum()

df_agrupado.to_excel('informe.xlsx')

df_pivote = df_zni.pivot_table(index= 'DEPARTAMENTO',columns= 'AÑO SERVICIO',values=['ENERGÍA ACTIVA'],aggfunc='sum')

df_pivote.to_excel('informe por año.xlsx')
```
